# displaying attribute

```python
import pandas as pd
Fire = pd.read_csv("Data\Fire_Data\\viirs-jpss1_2024_Portugal.csv")
Fire.head(0)
```

# checking Missing values

```python
Fire.isna().sum()
```

# Keeping Fires with high and normal confidence

```python
low = Fire.shape
Fire = Fire[ Fire["confidence"].str.lower() != "l" ].copy()
print(F"Dropped {low[0]-Fire.shape[0]} Fire with low confidence ")
```

# Keeping only Vegetation Fire

```python
all = Fire.shape
Fire = Fire[Fire["type"] == 0].copy()
print(F"Dropped {all[0]-Fire.shape[0]} Fire that are'nt presumed vegetation fire ")
```

# keeping Fire data from continental Portugal

```python
out = Fire.shape

Fire = Fire[
    (Fire['longitude'] > -9.55) & 
    (Fire['latitude'] > 36.9) & 
    (Fire['latitude'] < 42.2)
]

print(F"Dropped {out[0]-Fire.shape[0]} Fire that are'nt located in continental Portugal ")
```

```python
cols_to_keep = ['longitude','latitude']
Fire = Fire [cols_to_keep].copy()
Fire
```

```python
Fire.to_csv("Data\Fire_Data\\filtered_fire_data.csv", index=False)
```
