```python
import numpy as np
import pandas as pd
```

3. Chemins des fichiers


```python
csv_path_slope     = "/content/gdrive/MyDrive/Projet big data dossier organisé /dataTraiter/grid_with_fire_and_slope.csv"
csv_path_aspect    = "/content/gdrive/MyDrive/Projet big data dossier organisé /dataTraiter/grid_with_fire_aspect.csv"
csv_path_twi       = "/content/gdrive/MyDrive/Projet big data dossier organisé /dataTraiter/grid_with_fire_twi.csv"
csv_path_ndvi      = "/content/gdrive/MyDrive/Projet big data dossier organisé /dataTraiter/grid_with_fire_ndvi.csv"
csv_path_lst       = "/content/gdrive/MyDrive/Projet big data dossier organisé /dataTraiter/grid_with_LST.csv"
csv_path_distances = "/content/gdrive/MyDrive/Projet big data dossier organisé /dataTraiter/distances_grille_portugal.csv"
csv_path_elevation = "/content/gdrive/MyDrive/Projet big data dossier organisé /dataTraiter/elevation.csv"
output_csv_path    = "/content/gdrive/MyDrive/Projet big data dossier organisé /dataTraiter/datasetPaperIndian.csv"
```

4. Chargement des CSV


```python
df_slope     = pd.read_csv(csv_path_slope)
df_aspect    = pd.read_csv(csv_path_aspect)
df_twi       = pd.read_csv(csv_path_twi)
df_ndvi      = pd.read_csv(csv_path_ndvi)
df_lst       = pd.read_csv(csv_path_lst)
df_distances = pd.read_csv(csv_path_distances)
df_elevation = pd.read_csv(csv_path_elevation)
```

5. Aperçu des données


```python
print("--- Slope ---")
display(df_slope.head(5))

print("--- Aspect ---")
display(df_aspect.head(5))

print("--- TWI ---")
display(df_twi.head(5))

print("--- NDVI ---")
display(df_ndvi.head(5))

print("--- LST ---")
display(df_lst.head(5))

print("--- Distances ---")
display(df_distances.head(5))

print("--- Élévation ---")
display(df_elevation.head(5))
```

6. Fusion des datasets


```python
# Base : Slope
df_final = df_slope.copy()

# + TWI
df_final = pd.merge(df_final,
                    df_twi[['latitude', 'longitude', 'mean_twi']],
                    on=['latitude', 'longitude'], how='left')

# + Aspect
df_final = pd.merge(df_final,
                    df_aspect[['latitude', 'longitude', 'mean_aspect']],
                    on=['latitude', 'longitude'], how='left')

# + NDVI
df_final = pd.merge(df_final,
                    df_ndvi[['latitude', 'longitude', 'ndvi_final']],
                    on=['latitude', 'longitude'], how='left')

# + LST
df_final = pd.merge(df_final,
                    df_lst[['latitude', 'longitude', 'mean_LST_celsius']],
                    on=['latitude', 'longitude'], how='left')

# + Distances (routes, résidentiel)
df_final = pd.merge(df_final,
                    df_distances[['latitude', 'longitude', 'dist_residentiel_m', 'dist_route_m']],
                    on=['latitude', 'longitude'], how='left')

# + Élévation
df_final = pd.merge(df_final,
                    df_elevation[['latitude', 'longitude', 'Elevation']],
                    on=['latitude', 'longitude'], how='left')
```

7. Aperçu du dataset final


```python
print(f"Dimensions : {df_final.shape}")
print(f"Colonnes   : {list(df_final.columns)}")
display(df_final.head(5))
```

8. Sauvegarde


```python
df_final.to_csv(output_csv_path, index=False)
print(f"Dataset sauvegardé : {output_csv_path}")
```
